```python
import sys
sys.path.append("..")  # so we can import from src/

import pandas as pd
from src.config import INTERIM_DATA_DIR, PROCESSED_DATA_DIR
from src.features import build_feature_table

df = pd.read_parquet(INTERIM_DATA_DIR / "retail_clean.parquet")
df["Customer ID"] = df["Customer ID"].astype(str)

purchases = df[~df["is_cancellation"]].copy()
cancellations = df[df["is_cancellation"]].copy()

reference_date = df["InvoiceDate"].max()
print(f"Reference date: {reference_date}")
```

```python
features = build_feature_table(purchases, cancellations, reference_date)

print(f"Feature table shape: {features.shape}")
print(f"Churn rate: {features['is_churned'].mean():.1%}")
features.head()
```

```python
features.info()
features.isna().sum()
```

```python
PROCESSED_DATA_DIR.mkdir(parents=True, exist_ok=True)
out_path = PROCESSED_DATA_DIR / "customer_features.parquet"
features.to_parquet(out_path, index=False)
print(f"Saved to {out_path}")
```
